```python
# ==========================================================
# IMPHAL ROOFTOP SEGMENTATION USING U-NET
# ==========================================================

!pip install rasterio tensorflow scikit-learn -q

import os
import cv2
import numpy as np
import rasterio
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from tensorflow.keras import layers, Model

# ==========================================================
# MOUNT GOOGLE DRIVE
# ==========================================================

from google.colab import drive
drive.mount('/content/drive')

# ==========================================================
# FILE PATHS
# ==========================================================

RGB_PATH = "imphal_rgb.tif"
MASK_PATH = "imphal_mask.tif"

# ==========================================================
# LOAD TIFF FILES
# ==========================================================

with rasterio.open(RGB_PATH) as src:
    rgb = src.read([1, 2, 3])
    rgb = np.transpose(rgb, (1, 2, 0))

with rasterio.open(MASK_PATH) as src:
    mask = src.read(1)

print("RGB Shape:", rgb.shape)
print("Mask Shape:", mask.shape)

# ==========================================================
# NORMALIZATION
# ==========================================================

rgb = rgb.astype(np.float32)
rgb = rgb / np.max(rgb)

mask = (mask > 0).astype(np.float32)

# ==========================================================
# CREATE IMAGE TILES
# ==========================================================

IMG_SIZE = 256

images = []
masks = []

height, width = mask.shape

for i in range(0, height - IMG_SIZE, IMG_SIZE):
    for j in range(0, width - IMG_SIZE, IMG_SIZE):

        img_tile = rgb[i:i+IMG_SIZE, j:j+IMG_SIZE]
        mask_tile = mask[i:i+IMG_SIZE, j:j+IMG_SIZE]

        # Skip mostly empty tiles
        if np.mean(mask_tile) > 0.01:
            images.append(img_tile)
            masks.append(mask_tile)

images = np.array(images)
masks = np.array(masks)

masks = np.expand_dims(masks, axis=-1)

print("Total Tiles:", len(images))
print("Image Shape:", images.shape)
print("Mask Shape:", masks.shape)

# ==========================================================
# TRAIN TEST SPLIT
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    images,
    masks,
    test_size=0.2,
    random_state=42
)

print("Train:", X_train.shape)
print("Test:", X_test.shape)

# ==========================================================
# BUILD U-NET MODEL
# ==========================================================

def conv_block(x, filters):

    x = layers.Conv2D(filters, 3, padding="same",
                      activation="relu")(x)
    x = layers.Conv2D(filters, 3, padding="same",
                      activation="relu")(x)

    return x


def encoder_block(x, filters):

    c = conv_block(x, filters)
    p = layers.MaxPooling2D((2,2))(c)

    return c, p


def decoder_block(x, skip, filters):

    x = layers.Conv2DTranspose(
        filters,
        (2,2),
        strides=2,
        padding="same"
    )(x)

    x = layers.Concatenate()([x, skip])

    x = conv_block(x, filters)

    return x


def build_unet():

    inputs = layers.Input((256,256,3))

    s1, p1 = encoder_block(inputs, 64)
    s2, p2 = encoder_block(p1, 128)
    s3, p3 = encoder_block(p2, 256)

    b1 = conv_block(p3, 512)

    d1 = decoder_block(b1, s3, 256)
    d2 = decoder_block(d1, s2, 128)
    d3 = decoder_block(d2, s1, 64)

    outputs = layers.Conv2D(
        1,
        1,
        activation="sigmoid"
    )(d3)

    model = Model(inputs, outputs)

    return model

model = build_unet()

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# ==========================================================
# TRAIN MODEL
# ==========================================================

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=20,
    batch_size=8
)

# ==========================================================
# VISUALIZE RESULTS
# ==========================================================

index = np.random.randint(0, len(X_test))

sample_image = X_test[index]
sample_mask = y_test[index]

prediction = model.predict(
    np.expand_dims(sample_image, axis=0)
)[0]

prediction = (prediction > 0.5).astype(np.uint8)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(sample_image)
plt.title("RGB Image")

plt.subplot(1,3,2)
plt.imshow(sample_mask.squeeze(), cmap='gray')
plt.title("Ground Truth")

plt.subplot(1,3,3)
plt.imshow(prediction.squeeze(), cmap='gray')
plt.title("Prediction")

plt.show()

# ==========================================================
# SAVE MODEL
# ==========================================================
```
